```python
import ollama
```

```python
input_text = input("Enter text: ")


response = ollama.chat(
    model="llama3",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un esperto creatore di contenuti educativi. "
                "Il tuo compito è analizzare il testo fornito dall'utente e generare 3 domande di comprensione. "
                "Le domande devono testare la reale comprensione dei fatti e dei dettagli principali presenti nel testo. "
                "DEVI SCRIVERE LE DOMANDE TASSATIVAMENTE IN LINGUA ITALIANA. "
                "Elenca le domande in modo ordinato utilizzando un elenco numerato. "
                "Non aggiungere introduzioni o altri contenuti superflui; restituisci esclusivamente le domande."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
    options={
        "temperature": 0.6
    }
)

print(response['message']['content'])
```
