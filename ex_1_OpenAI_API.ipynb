**Traccia:** Scrivi un prompt che traduca un testo dall'italiano all'inglese e viceversa. Usa il prompt per tradurre una breve descrizione di un prodotto.

**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python
%pip install ollama
```

```python
import ollama

input_text = input("Inserisci il testo da tradurre: ")

response = ollama.chat(
    model="llama3",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un traduttore professionista bidirezionale (Italiano-Inglese). "
                "Il tuo UNICO compito è tradurre il testo fornito dall'utente. "
                "Se il testo dell'utente è in italiano, traducilo in inglese. "
                "Se è in inglese, traducilo in italiano. "
                "Fornisci SOLO la traduzione finale pulita. Non aggiungere saluti, introduzioni, spiegazioni o note."
            )
          },
        {
            "role": "user",
            "content": f"Traduci questo testo: {input_text}"
        }
    ],
)

print("\nRisultato traduzione:")
print(response['message']['content'])
```
