```python
import ollama

input_text = input("Enter text: ")

response = ollama.chat(
    model="llama3",
    format="json",  
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente specializzato nell'estrazione di dati da testi. "
                "Analizza la descrizione fornita dall'utente ed estrai esclusivamente tre informazioni: "
                "nome, età e professione. "
                "Restituisci unicamente un oggetto JSON con le chiavi 'nome', 'eta' e 'professione'. "
                "Se una delle informazioni non è presente nel testo, inserisci 'Non specificato' come valore. "
                "Non aggiungere introduzioni, commenti, note o preamboli esterni al JSON."
            )
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
    options={
        "temperature": 0.1  
    }
)

print(response['message']['content'])
```
