**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile.

```python
import ollama
```

```python
input_text = input("Enter text: ")
data = {"name": "Giovanni", "age": 30, "city": "Roma", "profession": "Ingegnere"};


response = ollama.chat(
    model="llama3",
    messages=[
        {
            "role": "system",
            "content": (
                "Sei un assistente specializzato nella generazione di testi a partire da dati strutturati. "
                "Il tuo compito è prendere il dizionario Python fornito dall'utente e trasformarlo in una descrizione testuale completa, fluida, naturale e grammaticalmente corretta sulla persona. "
                "DEVI RISPONDERE TASSATIVAMENTE IN LINGUA ITALIANA."
                "Non aggiungere commenti esterni, note o preamboli. Restituisci solo la descrizione discorsiva."
            )
        },
        {
            "role": "user",
            "content": f"Nota dell'utente: {input_text}\nEcco i dati strutturati da trasformare in testo: {data}"
        }
    ],
    options={
        "temperature": 0.3
    }
)

print(response['message']['content'])
```
